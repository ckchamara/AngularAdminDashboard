```python
# @title Run folder size report
# @markdown මේක run කරලා permission දුන්නම හරි. කැමති option තෝරගන්න.
Folder_Location = "" # @param {"type":"string","placeholder":"\"/content/drive/My Drive\""}
# @markdown මෙතනට size එක calculate කරන්න ඕන Folder එකේ location එක දාන්න. තියෙන විදිහට තිබ්බොත් මුළු drive එකේම ප්‍රමාණය calculate වෙයි.
OutputFile = "" # @param {"type":"string","placeholder":"\"/content/drive/MyDrive/folder_sizes.txt\""}
# @markdown මෙතනට text file එක හැදෙන්න ඕන තැනයි, නමයි දාන්න. තියෙන විදිහට තිබ්බොත් home folder එකේ folder_sizes.txt කියල හැදෙයි.

from google.colab import drive
drive.mount('/content/drive')

import os
import subprocess

BASE_PATH = Folder_Location

def human_readable_size(size_kb):
    size_bytes = size_kb * 1024
    for unit in ['B', 'KB', 'MB', 'GB', 'TB']:
        if size_bytes < 1024:
            return f"{size_bytes:.2f} {unit}"
        size_bytes /= 1024
    return f"{size_bytes:.2f} PB"

def list_dirs(path):
    try:
        return [os.path.join(path, name) for name in os.listdir(path)
                if os.path.isdir(os.path.join(path, name))]
    except Exception:
        return []

def get_folder_size_kb(path):
    try:
        result = subprocess.run(['du', '-s', '--block-size=1K', path],
                                stdout=subprocess.PIPE,
                                stderr=subprocess.PIPE,
                                text=True)
        if result.returncode == 0:
            size_kb = int(result.stdout.split()[0])
            return size_kb
    except Exception:
        pass
    return 0

# Get folder sizes at any level
def get_folder_sizes(path):
    subdirs = list_dirs(path)
    sizes = []
    for d in subdirs:
        size_kb = get_folder_size_kb(d)
        sizes.append((d, size_kb))
    sizes.sort(key=lambda x: x[1], reverse=True)
    return sizes

# Build the report
output_lines = []

# Level 1: top-level folders
level1 = get_folder_sizes(BASE_PATH)
output_lines.append("Top-level folders sorted by size:\n")
for path, size in level1:
    output_lines.append(f"{path} -> {human_readable_size(size)}")
output_lines.append("\n")

# Level 2 and 3
for top_path, _ in level1:
    output_lines.append(f"Subfolders in: {top_path}")
    level2 = get_folder_sizes(top_path)
    if not level2:
        output_lines.append("  (No subfolders)\n")
        continue

    for l2_path, l2_size in level2:
        output_lines.append(f"  {l2_path} -> {human_readable_size(l2_size)}")

        level3 = get_folder_sizes(l2_path)
        if level3:
            for l3_path, l3_size in level3:
                output_lines.append(f"    {l3_path} -> {human_readable_size(l3_size)}")
    output_lines.append("\n")

# Save output
output_file = OutputFile
with open(output_file, 'w') as f:
    f.write('\n'.join(output_lines))

print(f"Accurate folder size info saved to: {output_file}")
```
